Consider an array $A=[a_1,a_2,\ldots,a_n]$ containing $n$ elements. The only operation you are allowed is to check whether two elements are equal, i.e. "is $a_i$ equal to $a_j$?" and these will be the elementary operations.

In what follows, assume that $n$ is a power of 2, i.e. $n=2^k$ for some positive integer $k$.

An element $x$ is the *majority* element in $A$ if it occurs more than $n/2$ times in the array. So for example, $[1,2,1,1]$ (with $n=4$) has a majority element 1 because it occurs $3>n/2$ times. The array $[1,2,3,3]$ has no majority element: 3 occurs twice, but this isn't enough because it is not true that $2>n/2$.

We will consider two distinct algorithms for finding the majority element in a list.

**(a)** Let us first start with a naive approach. Describe, in words, what the different steps of the following algorithm do and derive the complexity of this algorithm in terms of elementary operations. **[ 5 ]**
    
```Python
def majority(A):
    n = len(A)
    for i in range(n):
        c = 0
        for j in range(n):
            if A[i]==A[j]:
                c = c+1
        if c>n/2:
            return A[i]
    return None
```

### Model Answer
**(a)**

For each element `A[i]`, the algorithm counts how many entries of `A` are equal to it (the inner loop over `j`, which uses `n` equality checks; the comparison of `A[i]` with itself is intended, since the element counts as one occurrence of itself). The counter `c` is reset for each `i`. After the inner loop finishes, if `c > n/2` then `A[i]` occurs more than `n/2` times, so it is the majority element and is returned. Otherwise the algorithm moves on to the next `i`. If no element passes the check, it returns `None`.

**Complexity:**

- The inner loop always performs exactly $n$ equality checks per value of `i`.
- In the worst case (no majority element, or the majority element is only found late), the outer loop runs $n$ times.
- Total number of elementary operations: $n \cdot n = n^2$.
- Best case: if `A[0]` is the majority element, the function returns after the first pass, costing only $n$ comparisons.

Therefore the worst-case complexity is $O(n^2)$.

<br>

**(b)**

**Explanation:**

- **Key claim:** if $x$ is the majority element of $A$, then it is the majority element of the left half or the right half (or both). Otherwise it would appear at most $n/4 + n/4 = n/2$ times in total, which is not more than $n/2$.
- **Divide:** split $A$ into a left half $L$ and a right half $R$, and recursively find the majority element of each (or `None` if there is none).
- **Combine**, depending on what the two recursive calls return:
  - Both return `None`: neither half has a majority, so $A$ has none. Return `None`.
  - Both return the same element $x$: $x$ appears more than $n/4 + n/4 = n/2$ times in $A$, so it is the majority. Return $x$ with no recount.
  - Only one half returns a candidate, or the two candidates differ: a majority of a half is only a *candidate* for the whole array, so count its occurrences in all of $A$. Return it if the count exceeds $n/2$; if both candidates are counted, return whichever passes, otherwise `None`.
- **Base case:** an array of one element has that element as its majority.

<br>

**(c)**


Let $T(n)$ be the number of equality checks for an array of size $n$.

- **Divide:** the array is split into two halves of size $n/2$ (slicing involves no equality checks).
- **Conquer:** two recursive calls are made, each on size $n/2$, costing $2T(n/2)$.
- **Combine:** in the worst case (the halves return different candidates), the `for` loop runs over all $n$ elements and does at most 2 equality checks per element (`num == L`, `num == R`), so at most $2n = O(n)$ checks. If $L = R$, no loop is run, which is cheaper.

This gives the recurrence

$$T(n) = 2\,T(n/2) + O(n)$$

so $a = 2$, $b = 2$, $d = 1$.

Since $\log_b a = \log_2 2 = 1 = d$, the middle case of the master method applies:

$$T(n) = O(n^d \log n) = O(n \log n)$$

This is better than the naive $O(n^2)$ algorithm from part (a).

## Python template

````python
def majority(x):
    if(not x):
        return None
    elif(len(x) == 1):
        return x[0]
    # floor division
    mid = len(x)//2
    L = majority(x[:mid])
    R = majority(x[mid:])
    cl = 0
    cr = 0
    
    if (L == R):
        return L
    
    for i in range(len(x)):
        if(x[i] == L):
            cl += 1
        elif(x[i] == R):
            cr += 1

    if(cl > len(x)/2):
        return x.index(L)
    elif(cr > len(x)/2):
        return x.index(R)
    else:
        return -1

x1 = [1,2,3,4,3,2,2,2,2,2,2,3,2,1,2,3]
x2 = [8,9,10,10]
maj1 = majority(x1)
maj2 = majority(x2)
print(maj1, maj2, x1[maj1])
if maj1==-1:
    print("Wrong! (1)")
elif maj1>=len(x1):
    print("Wrong! (2)")
elif x1[maj1]!=2:
    print("Wrong! (3)")
elif maj2!=-1:
    print("Wrong! (4)")
else:
    print("You got it!")
````

## C++ template

```c++
#include<iostream>
#include<vector>
#include<algorithm>

using namespace std;

// Returns an index into x of the majority element of x[lo..hi), or -1 if none.
int maj(const vector<int> &x, int lo, int hi)
{
    int n = hi - lo;
    if (n == 0) return -1;
    if (n == 1) return lo;

    int mid = lo + n / 2;
    int L = maj(x, lo, mid);   // candidate index from left half (or -1)
    int R = maj(x, mid, hi);   // candidate index from right half (or -1)

    // Same value is a majority of both halves, so it is a majority of the whole range
    if (L != -1 && R != -1 && x[L] == x[R]) return L;

    // Otherwise verify each candidate against the whole range
    int cl = 0, cr = 0;
    for (int i = lo; i < hi; i++) {
        if (L != -1 && x[i] == x[L]) cl++;
        if (R != -1 && x[i] == x[R]) cr++;
    }

    if (L != -1 && cl > n / 2) return L;
    if (R != -1 && cr > n / 2) return R;
    return -1;
}

int majority(vector<int> &x)
{
    return maj(x, 0, x.size());
}
```
